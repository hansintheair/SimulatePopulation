```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
def delta_deaths(birth_rate, death_chance, start_pop, p):
    
    if p == 0:
        return 0
    else:
        return 1 - (birth_rate - death_chance*start_pop)**(p-1)

def total_deaths(n_periods, birth_rate, death_chance, start_pop):

    return pd.Series([sum(delta_deaths(birth_rate, death_chance, start_pop, t) for t in range(1, i)) for i in range(1, n_periods)])

def total_births(n_periods, birth_rate):

    return pd.Series([(t-1)*birth_rate for t in range(1, n_periods)])

def population(total_briths, total_deaths):
    
    return total_births - total_deaths
```

```python
start_pop = 1
birth_rate = 1
death_chance = 0.1
n_periods = 50
```

```python
s_deaths = total_deaths(n_periods, birth_rate, death_chance, start_pop)
s_deaths[:10]
```
