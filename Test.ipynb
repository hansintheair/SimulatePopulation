```python
# Import required modules
import numpy as np
import matplotlib.pyplot as plt
from fractions import Fraction
import random
```

```python
def pop_growth(t, birth_rate, death_chance):
    # print(death_chance)
    # print(t*birth_rate)
    return  (death_chance**t) * (t* birth_rate)
```

```python
birth_rate = 1
death_chance = 0.5
N = 50
nums_a = [pop_growth(t, birth_rate, death_chance) for t in range(N)]
nums_b = [t for t in range(N)]
nums = [b- a for a, b in zip(nums_a, nums_b)]
```

```python
#Set up plot
fig, ax = plt.subplots(figsize=(8,5))
ax.set(
    title=f"Population Simulation",
    xlabel="Periods",
    ylabel="Population",
    xlim=(0, N)
)
ax.plot(nums)
ax.plot(nums_a)
ax.plot(nums_b)
```
