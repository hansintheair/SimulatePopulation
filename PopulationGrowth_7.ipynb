```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
def survive_exactly(death_chance, p):
    ''' Probability of surviving exactly to time step p given a constant 
        death_chance independent at each step.'''
        
    return (1-death_chance)**(p-1)*death_chance
```

Population growth when starting population is 0.

```python
population = [sum(survive_exactly(0.1, t) * (1/0.1) for t in range(1,i)) for i in range(1, 6) ]
population
```
