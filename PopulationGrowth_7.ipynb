in the previous part, we calculated the population using a recursive function. The function is conceptually easy to come up with, but it is inconvenient to use since it requires knowing the value of the previous iteration $P_{t-1}$. In order to calculate the population after $n$ time steps, we need to know the initial population $P_{t=0}$ and then recursively recalculate $P_t$ for $n$ periods.

In this part, we will begin to develop a function $f(t)$ that removes the need for this recursion, allowing us to calculate the exact population $P$ at any time $t$.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
def survive_exactly(death_chance, p):
    ''' Probability of surviving exactly to time step p given a constant 
        death_chance independent at each step.'''
    if not isinstance(p, int):
        raise TypeError
    else:
        return (1-death_chance)**(p-1)*death_chance if p > 0 else 0


def recursive_pop(n_periods, death_chance, birth_rate, pop):
    ''' Recursive function for popualtion growth with contstant birth rate and
        independent death chance'''

    p_function = [pop]
    for t in range(0,n_periods):  #pop_0 is first item, the function returns
                                    #pop_t+1 for any t.
        pop = pop*(1-death_chance)+birth_rate
        p_function.append(pop)
    return p_function
```

### Calculate Population with $f(t)$

If you look at the result of $Prob.(Survive \ At\ Least\ to\ t)$ from part 5 calculated for several time steps and then compare these to the result of the recursive function $P_t$ from the previous part (6), you will notice an interesting relationship: the values of $Prob.(Survive \ At\ Least\ to\ t)$ are smaller than the corresponding values of $P_t$ by exactly one decimal place.

```python
# Set simulation parameters
birth_rate = 1
death_chance = 0.1
n_periods = 10
```

```python
# Calculate P(survive at least to t) for n periods
p_lessequal_series = pd.Series(
    [
        sum(survive_exactly(death_chance, t)
        for t in range(0,i+1)) #range(0, 0) creates an empty list
        for i in range(0, n_periods+1)
    ],
    name="Prob.(surv. <= t)",
    index=(f"t={t:02}" for t in range(0, n_periods+1))
    )

# Calculate P_t for n periods
pop_recursive = pd.Series(
    recursive_pop(n_periods, death_chance, birth_rate, 0),
    name="Recursive P_t",
    index=(f"t={t:02}" for t in range(0, n_periods+1))
    )

pd.DataFrame.from_items([(p_lessequal_series.name, p_lessequal_series), (pop_recursive.name, pop_recursive)])
```
