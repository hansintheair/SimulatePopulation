```python
%matplotlib notebook

# Import required modules
import numpy as np
import matplotlib.pyplot as plt
from fractions import Fraction
import random
from matplotlib.animation import FuncAnimation
```

```python
def run_sim(n_periods):
    """ Run simulation of environment n times and plot the result"""

    #Set up plot
    fig, ax = plt.subplots(figsize=(8,5))
    ax.set(
        title="Simulation Outcome",
        xlabel="Periods",
        ylabel="Population",
        xlim=(0, n_periods+1),
        ylim=(0, 15)
    )
    l = [None for n in range(n_periods)]
    line = ax.plot(l)[0]

    plt.ion()
    fig.show()
    #Run simulation n times.
    for period in range(1, n_periods):
        l[period] = period
        line.set_ydata(l)
        fig.draw()


    # anim = FuncAnimation(fig, animate, interval=100)

# Run the simulation n times
run_sim(100)
```

```python
%matplotlib notebook
import numpy as np
import matplotlib.pyplot as plt

m = 100
n = 100
matrix = np.random.normal(0,1,m*n).reshape(m,n)

fig = plt.figure()
ax = fig.add_subplot(111)
plt.ion()

fig.show()
fig.canvas.draw()

for i in range(0,100):
    ax.clear()
    ax.plot(matrix[i,:])
    fig.canvas.draw()
```
